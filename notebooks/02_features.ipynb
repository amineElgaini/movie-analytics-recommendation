```python
import pandas as pd
import numpy as np

df = pd.read_pickle("../data/processed/movies_clean.pkl")
df.shape
```

```python
df["year"] = df["release_date"].dt.year
df["month"] = df["release_date"].dt.month
df["decade"] = (df["year"] // 10) * 10

df[["release_date", "year", "month", "decade"]].head()
print(df.head())
```

```python
df["n_genres"] = df["genres"].apply(len)
df["n_keywords"] = df["keywords"].apply(len)

df[["genres", "n_genres", "keywords", "n_keywords"]].head()
```

```python
def runtime_category(minutes):
    if pd.isna(minutes):
        return "unknown"
    elif minutes < 90:
        return "short"
    elif minutes <= 150:
        return "medium"
    else:
        return "long"

df["runtime_category"] = df["runtime"].apply(runtime_category)
df["runtime_category"].value_counts()
```

```python
df["overview_length"] = df["overview"].apply(lambda x: len(x.split()))

df["overview_length"].describe()
```

```python
from pathlib import Path

Path("../data/processed").mkdir(parents=True, exist_ok=True)
df.to_pickle("../data/processed/movies_features.pkl")
df.shape
```
